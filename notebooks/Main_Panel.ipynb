```python
%gui qt

from PySide6.QtWidgets import QApplication, QWidget, QVBoxLayout

from control_panel import (
    ControlPanel,
    DEVICES,
    HARDWARE_TYPES,
    create_hardware,
    DEFAULT_DEVICES
)

app = QApplication.instance()

if app is None:
    app = QApplication([])

app.setQuitOnLastWindowClosed(False)

hardware = create_hardware(DEVICES, HARDWARE_TYPES)

control_window = QWidget()
control_window.setWindowTitle("Hardware Control Panel")
control_window.setStyleSheet( "background: #07111f;")
layout = QVBoxLayout(control_window)

control_panel = ControlPanel(devices=DEVICES, hardware_types=HARDWARE_TYPES, hardware=hardware, default_devices=DEFAULT_DEVICES)

layout.addWidget(control_panel)

control_window.resize(1200,600)

control_window.show()

```
