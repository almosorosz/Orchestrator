```python
import time
import serial

SERIAL_PORT = "COM33"  # Double check that this matches your actual device port

# Configuration matching your GUI settings perfectly
serial_config = {
    "port": SERIAL_PORT,
    "baudrate": 9600,
    "bytesize": serial.EIGHTBITS,
    "parity": serial.PARITY_NONE,
    "stopbits": serial.STOPBITS_ONE,
    "timeout": 1.0
}

# The 'with' statement automatically opens and cleanly CLOSES the port
with serial.Serial(**serial_config) as ser:
    print(f"Successfully claimed {SERIAL_PORT}. Communicating...")
    
    # ---- 1. PING DEVICE ----
    ser.write(b"{M0101****}\r")
    time.sleep(0.200)  # Your mandatory 200ms delay from the image
    
    ping_reply = ser.read_until(b'\r').decode('ascii').strip()
    
    if ping_reply.startswith("{S0101"):
        print(f"✅ Status: CONNECTED (Device returned: {ping_reply})")
        
        # ---- 2. SET SETPOINT ----
        hex_val = "07D0"  # Example: 20.00 °C
        ser.write(f"{{M0100{hex_val}}}\r".encode('ascii'))
        time.sleep(0.200)  # 200ms delay again
        
        setpoint_reply = ser.read_until(b'\r').decode('ascii').strip()
        
        # ---- 3. VERIFY ----
        if setpoint_reply.startswith("{S0100"):
            print("🚀 Status: SETPOINT OK")
        else:
            print(f"❌ Status: SETPOINT FAILED (Received: {setpoint_reply})")
            
    else:
        print(f"❌ Status: OFFLINE or unexpected reply: '{ping_reply}'")

print("🔒 Port safely released. Your kernel will not crash.")
```

```python
ser.close()
```
